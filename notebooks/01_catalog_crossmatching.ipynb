# Catalog Cross-Matching

This notebook demonstrates cross-matching astronomical catalogs from different surveys (GMRT radio and SDSS optical data).

## Overview
- Load catalogs from GMRT (radio) and SDSS (optical) surveys
- Match objects based on sky coordinates
- Save matched results for further analysis

```python
import numpy as np
import sys
sys.path.append('../src')

from crossmatch import crossmatch, save_matches
```

## Load Catalogs

Load the GMRT radio catalog and SDSS optical catalog.

```python
# Load GMRT radio catalog (RA, Dec coordinates)
cat1 = np.genfromtxt('../data/raw/gmrt.csv', delimiter=',', skip_header=55, usecols=[5, 6], max_rows=5434)

# Load SDSS optical catalog (RA, Dec coordinates)
cat2 = np.genfromtxt('../data/raw/opticaldata.csv', delimiter=',', skip_header=1, usecols=[1, 2], max_rows=252198)

print(f"GMRT catalog: {len(cat1)} sources")
print(f"SDSS catalog: {len(cat2)} sources")
```

## Perform Cross-Matching

Match sources within 5 degrees (maximum distance threshold).

```python
# Cross-match with 5 degree maximum distance
matches, no_matches, time_taken = crossmatch(cat1, cat2, max_dist=5)

print(f"Matches found: {len(matches)}")
print(f"Unmatched sources: {len(no_matches)}")
print(f"Time taken: {time_taken:.2f} seconds")
```

## Save Results

```python
# Save matched catalog
save_matches(matches, '../data/results/matched_catalogs.txt')
print("Results saved to data/results/matched_catalogs.txt")
```

## Example Matches

Display the first few matches:

```python
print("First 10 matches (cat1_idx, cat2_idx, distance):")
for i, match in enumerate(matches[:10]):
    print(f"{i+1}. Index {match[0]} matched with index {match[1]}, distance: {match[2]:.4f} degrees")
```
